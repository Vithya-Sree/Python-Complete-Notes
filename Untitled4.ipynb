**Why do we use NumPy arrays instead of Python lists for data analysis?**

NumPy arrays are designed for numerical computation. They support fast vectorized operations, use memory efficiently, and allow mathematical operations to be performed on entire arrays without writing explicit loops.

```python
import numpy as np

# Python List
python_list = [10, 20, 30, 40, 50]

# NumPy Array
numpy_array = np.array([10, 20, 30, 40, 50])

print("Python List:", python_list)
print("NumPy Array:", numpy_array)

# Multiplication
print("\nPython List * 2:", python_list * 2)
print("NumPy Array * 2:", numpy_array * 2)

# Check types
print("\nList Type:", type(python_list))
print("Array Type:", type(numpy_array))
```
