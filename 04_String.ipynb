**String Methods**

```python
name = "hello world"
```

```python
#upper() — converts to uppercase
name.upper()
```

```python
#lower() — converts to lowercase
name.lower()
```

```python
#capitalize() — capitalizes the first character
name.capitalize()
```

```python
#title() — capitalizes the first letter of each word
name.title()
```

**Difference Between capitalize() and title()**

Both methods change the capitalization of a string, but they work differently.

**capitalize()**

capitalize() makes only the first character of the entire string uppercase and makes the rest lowercase.

**title()**

title() makes the first character of each word uppercase.

```python
#strip() — removes spaces from the beginning and end
name = "  hello  "
name.strip()
```

```python
#replace() — replaces part of a string
name.replace("hello", "hi")
```

```python
#split() — splits a string into a list
name = "hello world"
name.split()
```

```python
#join() — joins elements into a string
words = ["Hello", "World"]
" ".join(words)
```

```python
#find() — finds the position of a substring
name.find("world")
```

```python
#count() — counts how many times something occurs
"banana".count("a")
```

```python
#startswith() — checks whether a string starts with something
"hello".startswith("he")
```

```python
#endswith() — checks whether a string ends with something
"hello".endswith("lo")
```

```python
#isdigit() — checks whether all characters are digits
"123".isdigit()
```

```python
#isalpha() — checks whether all characters are alphabetic
"Hello".isalpha()
```

```python
#isalnum() — checks whether all characters are letters or numbers
"Hello123".isalnum()
```

**Important point**

String methods do not usually change the original string, because Python strings are immutable.

**Case Study: Student Registration System**

```python
student = "  john doe, john.doe@gmail.com, computer science  "

# Remove spaces
student = student.strip()

# Split data
data = student.split(",")

# Clean each field
name = data[0].strip()
email = data[1].strip()
department = data[2].strip()

# Format data
name = name.title()
department = department.title()

# Display
print("Name:", name)
print("Email:", email)
print("Department:", department)

# Validate email
if email.endswith("@gmail.com"):
    print("Email: Valid Gmail")

# Student ID validation
student_id = "20260125"

if student_id.isdigit():
    print("Student ID: Valid")
```

**Case Study: Online Shopping Order**

```python
order = "  laptop, electronics, 2, delivered  "

# Remove extra spaces
order = order.strip()

# Separate information
data = order.split(",")

# Clean each value
product = data[0].strip()
category = data[1].strip()
quantity = data[2].strip()
status = data[3].strip()

# Format
product = product.title()
category = category.title()

# Display
print("Product:", product)
print("Category:", category)
print("Quantity:", quantity)
print("Status:", status)

# Validate quantity
if quantity.isdigit():
    print("Quantity is valid")

# Check status
if status.lower() == "delivered":
    print("Order has been delivered")

# Search product
if product.lower().find("lap") != -1:
    print("Laptop found")
```

The find() method has an important rule:

If it finds the text → returns its position (0, 1, 2, etc.)

If it doesn't find the text → returns -1
